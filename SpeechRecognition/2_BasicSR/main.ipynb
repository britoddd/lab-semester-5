```python
from IPython.display import Audio, display

AUDIO_PATH = './sample.wav'

display(Audio(AUDIO_PATH))
```

```python
import soundfile as sf

signal, sample_rate = sf.read(AUDIO_PATH)

print("Signal:", signal)
print("Sample Rate:", sample_rate)
print("Info:", sf.info(AUDIO_PATH))
```

```python
# Tambahan aja
sf.write("output/sf_export.mp3", signal, 16000)  # kalo sample rate terlalu kecil dibanding aslinya, bakal jadi kayak dislow motion
```

```python
# Read, write audio file format .wav
import wave

audio = wave.open(AUDIO_PATH, 'r')

print("Channel:", audio.getnchannels())
print("Sample Width (byte):", audio.getsampwidth())
print("Sample rate:", audio.getframerate())
print("Total sample:", audio.getnframes())
print("Parameter:", audio.getparams())

frames = audio.readframes(audio.getnframes())

audio.close()
```

```python
audio = wave.open("output/wave_export.wav", 'w')
audio.setnchannels(1)
audio.setsampwidth(2)
audio.setframerate(44100)
audio.writeframes(frames)
```

```python
from pydub import AudioSegment

AudioSegment.from_wav(AUDIO_PATH)

def volume_manipulation():
    audio = AudioSegment.from_wav(AUDIO_PATH)
    audio = audio + 10
    audio.export("output/volume.wav", format="wav")
    
volume_manipulation()
```

```python
def repeat_audio():
    audio = AudioSegment.from_wav(AUDIO_PATH)
    audio = audio * 2
    audio.export("output/repeat.wav", format="wav")
    
repeat_audio()
```

```python
def reverse_audio():
    audio = AudioSegment.from_wav(AUDIO_PATH)
    audio = audio.reverse()
    audio.export("output/repeat.wav", format="wav")
    
reverse_audio()
```

```python
import librosa
signal, sample_rate = librosa.load(AUDIO_PATH, sr=sample_rate)
```

```python
import numpy as np

# White noise
def white_noise(signal, noise_rate):
    noise = np.random.normal(0, signal.std(), signal.size)
    augmented_signal = signal + noise * noise_rate
    return augmented_signal

augmented_signal = white_noise(signal, 0.5)
sf.write("output/white_noise.wav", augmented_signal, sample_rate)
```

```python
def time_stretch(signal, stretch_rate):
    return librosa.effects.time_stretch(signal, rate = stretch_rate)

augmented_signal = time_stretch(signal, 0.5)
sf.write("output/time_stretch.wav", augmented_signal, sample_rate)
```

```python
def pitch_shift(signal, sr, step):
    return librosa.effects.pitch_shift(signal, sr=sr, n_step=step)

augmented_signal = pitch_shift(signal, sample_rate, 10)
sf.write("output/pitch_shift.wav", augmented_signal, sample_rate)
```

```python
def random_gain(signal, min_gain, max_gain):
    gain_factor = np.random.uniform(min_gain, max_gain)
    return signal * gain_factor

augmented_signal = random_gain(signal, 10, 50)
sf.write("output/random_gain.wav", augmented_signal, sample_rate)
```
